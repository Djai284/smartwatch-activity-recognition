```python
import torch
import coremltools as ct
```

```python

model = torch.load("model.pt")
model.eval()
# Adjust the input shape to match your model requirements (e.g., batch size, channels, sequence length)
example_input = torch.randn(1, 6, 500)
traced_model = torch.jit.trace(model, example_input)
mlmodel = ct.convert(traced_model, inputs=[ct.TensorType(shape=example_input.shape)])
mlmodel.save("MyModel.mlmodel")
```
